```python
import os
import librosa
import soundfile as sf

# Путь к исходным файлам
input_folder = "/Users/nikitavdovichev/Documents/5 курс/ПроектнаяПрактика/dataset/audiofiles"

# Путь к папке, куда будут сохранены преобразованные аудиофайлы
output_folder = "/Users/nikitavdovichev/Documents/5 курс/ПроектнаяПрактика/dataset/audiofiles_16khz_mono"

# Создать папку для выходных файлов, если она не существует
os.makedirs(output_folder, exist_ok=True)

# Функция для преобразования аудиофайла
def convert_audio(input_path, output_path):
    # Загрузеп аудиофайла с исходной частотой дискретизации
    audio_data, original_sr = librosa.load(input_path, sr=None, mono=True)
    
    # Преобразование частоты дискретизации в 16 кГц
    audio_data_16khz = librosa.resample(audio_data, orig_sr=original_sr, target_sr=16000)
    
    # Сохранение преобразованного аудиофайла
    sf.write(output_path, audio_data_16khz, 16000, subtype='PCM_16')

# Обработка всех аудиофайлов в папке
for root, dirs, files in os.walk(input_folder):
    for file in files:
        # Проверка что файл является аудиофайлом
        if file.endswith((".wav", ".mp3", ".ogg", ".flac")):
            input_path = os.path.join(root, file)
            relative_path = os.path.relpath(input_path, input_folder)
            output_path = os.path.join(output_folder, relative_path)
            
            # Создать папку для выходных файлов, если она не существует
            os.makedirs(os.path.dirname(output_path), exist_ok=True)
            
            print(f"Обработка: {input_path} -> {output_path}")
            convert_audio(input_path, output_path)

print("Все файлы обработаны!")
```

```python
# Путь к папке с аудиофайлами
audio_folder = "/Users/nikitavdovichev/Documents/5 курс/ПроектнаяПрактика/dataset/audiofiles_16khz_mono"

# Инициализируем переменную для хранения общей длительности
total_duration = 0.0

# Проходим по всем файлам в папке для опредедления общей длины аудио
for filename in os.listdir(audio_folder):
    if filename.endswith(".wav"):  # Проверяем, что файл имеет расширение .wav
        file_path = os.path.join(audio_folder, filename)
        # Загружаем аудиофайл с помощью librosa
        audio, sr = librosa.load(file_path, sr=None)
        # Вычисляем длительность в секундах
        duration = librosa.get_duration(y=audio, sr=sr)
        # Добавляем длительность к общей сумме
        total_duration += duration

# Выводим общую длительность
print(f"Общая продолжительность аудиофайлов: {total_duration:.2f} секунд")
```
