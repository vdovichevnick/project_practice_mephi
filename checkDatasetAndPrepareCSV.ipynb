```python
import matplotlib.pyplot as plt
import librosa
import pandas as pd
import numpy as np
import sys
import os

metadata_csv = r"/Users/nikitavdovichev/Documents/5 курс/ПроектнаяПрактика/dataset/UrbanSound8K.csv"

df = pd.read_csv(metadata_csv)

# Выводим уникальные названия классов
unique_classes = df['class'].unique()
print(unique_classes)

df.head(20)
```

```python
import pandas as pd
import os
from IPython.display import Audio, display

class_search = 'gun_shot'

# Фильтруем DataFrame по классу
filter_by_class_df = df[df['class'] == class_search]

# Выбираем 3 случайных slice_file_name
random_slice_file_names = filter_by_class_df['slice_file_name'].sample(n=3)

# Путь к  папке с аудиофайлами
base_path = r"/Users/nikitavdovichev/Documents/5 курс/ПроектнаяПрактика/dataset/audiofiles_16khz_mono"

# Воспроизведение аудиофайлов
for slice_file_name in random_slice_file_names:
    audio_path = os.path.join(base_path, slice_file_name)
    audio = Audio(filename=audio_path)
    display(audio)
    print(f"Файл: {slice_file_name}")
```

```python
# Добавляем столбец is_gun_shot
df['is_gun_shot'] = (df['class'] == 'gun_shot').astype(int)

# Путь для сохранения нового CSV файла
new_metadata_csv = r"/Users/nikitavdovichev/Documents/5 курс/ПроектнаяПрактика/dataset/UrbanSound8K_edited.csv"

# Сохраняем DataFrame в новый CSV файл
df.to_csv(new_metadata_csv, index=False)

# Подсчитываем количество вхождений каждого класса
class_counts = df['class'].value_counts()

# Выводим результат
print("Количество вхождений каждого класса:")
print(class_counts)

# Подсчитываем количество вхождений каждого значения в столбце is_gun_shot
is_gun_shot_counts = df['is_gun_shot'].value_counts()

# Выводим результат
print("\nКоличество вхождений каждого значения в столбце is_gun_shot:")
print(is_gun_shot_counts)

# Выводим первые несколько строк DataFrame для проверки
print(df.head())
```
